```python

studies = [
    {"id": 1, "title": "Cancer Biomarkers", "status": "active"},
    {"id": 2, "title": "Protein Folding", "status": "completed"},
    {"id": 3, "title": "Drug Discovery", "status": "active"}
]

def get_study_by_id(studies, study_id):
    for study in studies:
        if study["id"] == study_id:
            return study
    return None

#testing
print(get_study_by_id(studies, 2))
# Expected: # {"id": 2, "title": "Protein Folding", "status": "completed"}

print(get_study_by_id(studies, 99))
# Expected: None
#time complexity O(n)
#space complexity O(1)
```

```python

def get_studies_by_status(studies, status):
    # Your implementation
    result = []
    for study in studies:
        if study["status"] == status:
            result.append(study)
    return result

print(get_studies_by_status(studies, "active"))
# Expected: studies with IDs 1 and 3

print(get_studies_by_status(studies, "paused"))
# Expected: []
#time complexity O(n) because we are iterating through the list of studies once
#space complexity O(k) where k is the number of studies with the specified status
```
